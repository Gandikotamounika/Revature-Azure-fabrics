```python
# Q1. Given a product price of 12,500, apply a discount of 10% and then add 18% GST. Calculate the final payable amount.
price = 12500

discount = price * 10 / 100
price_after_discount = price - discount

gst = price_after_discount * 18 / 100

final_amount = price_after_discount + gst

print("Final Payable Amount =", final_amount)

```

```python
#Q11. Iterate through a list of orders and count how many orders are above 1,00,000.
orders = [45000, 120000, 98000, 150000, 200000]

count = 0

for order in orders:
    if order > 100000:
        count += 1

print("Total Orders Above 1,00,000 =", count)
```

```python
# Q12. Simulate payment retry logic where the system retries payment a maximum of 3 times if payment fails.
attempt = 1

while attempt <= 3:
    print("Payment Attempt", attempt)

    payment_success = False

    if payment_success:
        print("Payment Successful")
        break

    attempt += 1

if attempt > 3:
    print("Payment Failed After 3 Attempts")

```

```python
#Q13. Write a program that keeps checking order status until status becomes 'Delivered' or maximum 5 attempts are reached.
status = "Processing"
attempt = 1

while status != "Delivered" and attempt <= 5:
    print("Checking Order Status... Attempt", attempt)

    attempt += 1

print("Maximum Attempts Reached")

```

```python
#Q14. Inventory restock check: Keep checking stock level every iteration and stop when stock becomes greater than 100.
stock = 40

while stock <= 100:
    print("Current Stock =", stock)

    stock += 20

print("Stock Level Sufficient")

```

```python
#Q15. You are given a list of orders with order_id and amount.
#- Skip cancelled orders
#- Classify remaining orders as Premium or Regular
#- Count total premium orders

Use operators, IF–ELSE, FOR loop, and counters.

orders = [
    {"order_id": 101, "amount": 120000, "status": "Completed"},
    {"order_id": 102, "amount": 45000, "status": "Cancelled"},
    {"order_id": 103, "amount": 75000, "status": "Completed"},
    {"order_id": 104, "amount": 150000, "status": "Completed"}
]

premium_count = 0

for order in orders:

    if order["status"] == "Cancelled":
        continue

    if order["amount"] > 100000:
        print(order["order_id"], "- Premium Order")
        premium_count += 1
    else:
        print(order["order_id"], "- Regular Order")

print("Total Premium Orders =", premium_count)
```
