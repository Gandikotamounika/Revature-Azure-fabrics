```python
#Problem 1 — Write Products into CSV File
import csv

products = [
    [101, "Laptop", "Electronics", 75000, 10],
    [102, "Mouse", "Accessories", 800, 50],
    [103, "Keyboard", "Accessories", 1500, 30]
]

with open("products.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Writing header
    writer.writerow(["ID", "Name", "Category", "Price", "Quantity"])

    # Writing product data
    writer.writerows(products)

print("Products written into CSV file successfully")
```

```python
#Problem 2 — Read Data from CSV File
import csv

with open("products.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
#3 Problem 3 — Delete Product from CSV File
import csv

updated_products = []

with open("products.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        updated_products.append(row)

# Remove product with ID 103
filtered_products = []

for row in updated_products:
    if row[0] != "103":
        filtered_products.append(row)

# Rewrite file
with open("products.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(filtered_products)

print("Product deleted successfully")
```

```python
#4 — Display Products with Price Greater than 5000
import csv

with open("products.csv", "r") as file:
    reader = csv.reader(file)

    next(reader)

    print("Products with price greater than 5000:")

    for row in reader:
        if int(row[3]) > 5000:
            print(row)
```

```python
#5 — Write Products into JSON File
import json

products = [
    {"id": 101, "name": "Laptop", "category": "Electronics", "price": 75000, "quantity": 10},
    {"id": 102, "name": "Mouse", "category": "Accessories", "price": 800, "quantity": 50},
    {"id": 103, "name": "Keyboard", "category": "Accessories", "price": 1500, "quantity": 30},
    {"id": 104, "name": "Mobile", "category": "Electronics", "price": 25000, "quantity": 15},
    {"id": 105, "name": "Headphones", "category": "Accessories", "price": 2000, "quantity": 20}
]

with open("products.json", "w") as file:
    json.dump(products, file, indent=4)

print("Products written into JSON file successfully")
```

```python
#6 — Read Data from JSON File
import json

with open("products.json", "r") as file:
    products = json.load(file)

for product in products:
    print(product)
```

```python
#7 — Delete Product from JSON File
import json

with open("products.json", "r") as file:
    products = json.load(file)

updated_products = []

for product in products:
    if product["id"] != 102:
        updated_products.append(product)

with open("products.json", "w") as file:
    json.dump(updated_products, file, indent=4)

print("Product deleted successfully")
```

```python
#8 — Update Product Quantity
import json

with open("products.json", "r") as file:
    products = json.load(file)

for product in products:
    if product["name"] == "Laptop":
        product["quantity"] = 25

with open("products.json", "w") as file:
    json.dump(products, file, indent=4)

print("Laptop quantity updated successfully")
```

```python
#9 — Display Electronics Products Only
import json

with open("products.json", "r") as file:
    products = json.load(file)

print("Electronics Products:")

for product in products:
    if product["category"] == "Electronics":
        print(product)
```

```python
#10 — Append New Product to CSV
import csv

new_product = [106, "Monitor", "Electronics", 12000, 8]

with open("products.csv", "a", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(new_product)

print("New product appended to CSV file")
```

```python
#11 — Append New Product to JSON
import json

new_product = {
    "id": 106,
    "name": "Monitor",
    "category": "Electronics",
    "price": 12000,
    "quantity": 8
}

with open("products.json", "r") as file:
    products = json.load(file)

products.append(new_product)

with open("products.json", "w") as file:
    json.dump(products, file, indent=4)

print("New product appended to JSON file")
```
